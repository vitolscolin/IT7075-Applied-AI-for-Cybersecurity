## 8. Computational inventory (§3.8) — run on EACH platform
Run this self-contained cell locally, in Colab, and in the remote VS Code kernel if Jetstream2 is available. Colab is detected automatically; set `PLATFORM_LABEL = "Jetstream2"` on the remote VM and retain each platform's output (separate notebook copies or screenshots). It uses system utilities **and** a framework query. Missing GPU hardware is a valid result, not a failed lab; a missing utility alone does not prove the hardware is absent.

`psutil` and `torch` are in the lab requirements. In Colab run §5's self-contained dependency cell first; it does not need any local requirements file. GPU-enabled PyTorch installation may depend on your host's driver; use the [official installer selector](https://pytorch.org/get-started/locally/) if you need a different build.

````python
import platform
import shutil
import subprocess
import sys
import psutil
import torch

from pathlib import Path
PLATFORM_LABEL = "Colab" if Path("/content").is_dir() and "google.colab" in sys.modules else "Local"
# On the remote VM, change this to: PLATFORM_LABEL = "Jetstream2"
print("Platform:", PLATFORM_LABEL)
print("OS:", platform.platform())
print("Python:", sys.executable)
print("CPU:", platform.processor() or "See system utility output below")
print("Physical cores:", psutil.cpu_count(logical=False))
print("Logical CPUs:", psutil.cpu_count(logical=True))
print("RAM GiB:", round(psutil.virtual_memory().total / 1024**3, 2))

def show_command(args):
    print("\n$", " ".join(args))
    if not shutil.which(args[0]):
        print("Utility unavailable; use your OS hardware interface for additional evidence.")
        return
    try:
        proc = subprocess.run(args, capture_output=True, text=True, timeout=30)
        print(proc.stdout or proc.stderr)
        print("Exit status:", proc.returncode)
    except subprocess.TimeoutExpired:
        print("Utility timed out; inspect hardware through the OS interface.")

if platform.system() == "Linux":
    show_command(["lscpu"])
    show_command(["free", "-h"])
    show_command(["lspci"])
elif platform.system() == "Darwin":
    show_command(["system_profiler", "SPHardwareDataType", "SPDisplaysDataType"])
else:
    show_command(["powershell", "-NoProfile", "-Command",
                  "Get-CimInstance Win32_Processor | Select-Object Name,NumberOfCores,NumberOfLogicalProcessors; "
                  "Get-CimInstance Win32_ComputerSystem | Select-Object TotalPhysicalMemory; "
                  "Get-CimInstance Win32_VideoController | Select-Object Name,AdapterRAM"])
show_command(["nvidia-smi"])
print("\nPyTorch:", torch.__version__)
print("PyTorch CUDA build:", torch.version.cuda)
print("torch.cuda.is_available():", torch.cuda.is_available())
print("CUDA device count:", torch.cuda.device_count())
for i in range(torch.cuda.device_count()):
    print("CUDA device", i, ":", torch.cuda.get_device_name(i))
mps = getattr(torch.backends, "mps", None)
print("torch.backends.mps.is_available():", bool(mps and mps.is_available()))
````
